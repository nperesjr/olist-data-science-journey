```python
from pathlib import Path
import pandas as pd

pasta = Path("../data")
tabelas = {}

for arquivo in sorted(pasta.glob("*.csv")):
    nome = arquivo.stem.replace("olist_", "").replace("_dataset", "")
    tabelas[nome] = pd.read_csv(arquivo)

    #print(nome, tabelas[nome].shape)
    #display(tabelas[nome].head())


print(tabelas.keys())
```

```python
tabelas["orders"].info() #colunas, tipos, valores não nulos
tabelas["orders"].nunique() #valores diferentes em cada coluna
tabelas["orders"].isna().sum() #valores vazios em cada coluna
```

```python
for nome, df in tabelas.items():
    if "order_id" in df.columns:
        print(nome, "linhas:", len(df), "| pedidos diferentes:", df["order_id"].nunique())
```

```python
pedidos = tabelas["orders"]
itens = tabelas["order_items"]

sem_item = pedidos[~pedidos["order_id"].isin(itens["order_id"])]

print(len(sem_item))
sem_item["order_status"].value_counts()
```
